```python
import pandas as pd
data = {
    "Hours": [2, 4, None, 8, 10, 12, None, 16],
    "Marks": [35, 45, 50, None, 75, 85, 90, None]
}
df=pd.DataFrame(data)
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df_clean= df.dropna()
print(df_clean)
```

```python
df['Hours']= df['Hours'].fillna(df['Hours'].mean())
print(df['Hours'])
```

```python
df['Marks']= df['Marks'].fillna(df['Marks'].mean())
print(df['Marks'])
```
